## 1. Why Python and NumPy?

- Python = orchestration language for AI
- NumPy = fast numerical computation
- ML frameworks (PyTorch, TensorFlow) are NumPy-like
- Understanding NumPy → understanding tensors later

Key mindset shift:
- Scalar → Vector → Matrix → Tensor


```python
# Python scalar
x = 3
y = 4
x * y
```

```python
import numpy as np

# NumPy array
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

a * b
```

```python
# Common ways to create vectors
np.zeros(5)
```

```python
np.arange(0, 10, 2)
```

```python
np.linspace(0, 1, 5)
```

## 2. Shape and Dimensionality

- shape → structure
- ndim → number of dimensions
- size → total number of elements


```python
v = np.array([1, 2, 3])

v.shape, v.ndim, v.size
```

```python
M = np.array([[1, 2, 3],
              [4, 5, 6]])

M.shape, M.ndim, M.size
```

## 3. Vectorization

Vectorization:
- Eliminates explicit loops
- Faster and more readable
- Critical for AI performance


```python
# Loop-based (not recomm for ML)
result = []
for i in range(len(a)):
    result.append(a[i] * b[i])

result
```

```python
# Vectorized
a * b
```

## 4. Broadcasting

Broadcasting allows NumPy to operate on arrays of different shapes.


```python
v = np.array([1, 2, 3])
v + 10
```

```python
M = np.array([[1, 2, 3],
              [4, 5, 6]])

M + v
```

```python
x = np.array([1, 2, 3, 4, 5])

x.sum(), x.mean(), x.std()
```

```python
M.sum(axis=0)  # column-wise
```

```python
M.sum(axis=1)  # row-wise
```

```python
x = np.array([10, 20, 30, 40, 50])

x[0], x[-1]
```

```python
x[1:4]
```

```python
M[0, :]
```

```python
M[:, 1]
```
