## 1. Why Visualization Matters

Visualization helps you:
- Understand data distributions
- Detect outliers and errors
- Monitor training progress
- Debug loss, gradients, and predictions

If you can’t visualize it, you probably don’t understand it yet.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)
y = x ** 2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("A Basic Line Plot")
plt.show()
```

```python
# Scatter Plot
x = np.random.randn(100)
y = 2 * x + np.random.randn(100) * 0.5

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Scatter Plot")
plt.show()
```

```python
# Histogram
data = np.random.randn(1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

```python
# Multi-plot figure
x = np.linspace(0, 10, 100)

plt.plot(x, x, label="y = x")
plt.plot(x, x**2, label="y = x^2")
plt.plot(x, np.sqrt(x), label="y = sqrt(x)")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.title("Multiple Lines")
plt.show()
```

```python
# Visualizing loss During ML model Training
losses = []

w = 0.0
lr = 0.01

for _ in range(50):
    loss = (w - 3)**2
    losses.append(loss)
    w = w - lr * 2*(w - 3)

plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss Curve")
plt.show()
```

```python
## Contour Plot
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

plt.contour(X, Y, Z)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Contour Plot")
plt.show()
```

## 2. Common Mistakes

- Forgetting labels and titles
- Misleading scales
- Overplotting without clarity
- Ignoring distribution shape

Visualization should clarify, not decorate.


## 3. Key Takeaways

- Visualization builds intuition
- Line plots show trends
- Scatter plots show relationships
- Histograms show distributions
- Contours show optimization landscapes
